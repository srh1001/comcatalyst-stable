# **Load results.**

```python
# load from json
res = load_results_from_dir("../../data/xpr_smt/xpr_smt_case04_bo30")
```

```python
# restrict
res = [r for r in res if (r["hyperparams"]["seed"]<10) or (r["hyperparams"]["seed"]>14 and r["hyperparams"]["kernel"]=="EXP_HOMO_HSPHERE")]
res = [r for r in res if r["hyperparams"]["sigma_noise"]<1]
```

```python
# restrict number of points
res = slice_results_at_n_points(results=res, n_points=35)
```

```python
# check
print("Total runs:", len(res))
for h in res[0]["hyperparams"]:
    values = sorted(set(r["hyperparams"][h] for r in res))
    print(f"{h:<27}: {'':>1}{values}")
```

# **Plot results.**

## **Convergence curve.**

```python
# convergence
for n_init in [2, 6, 10, 14]:
    for sigma_noise in [0.1, 0.5]:
        for bo_criterion in ["EI", "LCB"]:
            for kernel in ["GOWER", "CONT_RELAX", "EXP_HOMO_HSPHERE"]:
                print(f"n_init={n_init}, sigma_noise={sigma_noise}, bo_criterion={bo_criterion}, kernel={kernel}")
                plot_convergence_curve(
                    results=res,
                    metric="y_true_at_x_data",
                    filters={"n_init": n_init, 
                             "sigma_noise": sigma_noise,
                             "bo_criterion": bo_criterion, 
                             "kernel": kernel
                            },
                    ylim=(-6,6),
                    figsize=(9,5),
                    vline=30,
                    fontsize=12
                )
```
